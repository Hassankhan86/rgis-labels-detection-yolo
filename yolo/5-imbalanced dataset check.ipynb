```python
from pathlib import Path
from collections import Counter

root = Path('dataset3')
classes = [l.strip() for l in (root/'classes.txt').read_text().splitlines() if l.strip()]
counts = Counter()
for split in ('train', 'valid', 'test'):
    for lp in (root/'labels'/split).iterdir():
        for line in lp.read_text().splitlines():
            if line.strip():
                counts[int(line.split()[0])] += 1

for cid, name in enumerate(classes):
    print(name, counts.get(cid, 0))
```
